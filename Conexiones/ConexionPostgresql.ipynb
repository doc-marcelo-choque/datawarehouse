```python
from sqlalchemy import create_engine

usuario = "postgres"
password = "12345"
servidor = "localhost"
puerto = "5432"
base_datos = "tienda_friki"

cadena = (
    f"postgresql+psycopg2://"
    f"{usuario}:{password}@"
    f"{servidor}:{puerto}/"
    f"{base_datos}"
)

print(cadena)

engine = create_engine(cadena)

try:
    with engine.connect() as conexion:
        print("Conexión exitosa")
except Exception as error:
    print(type(error))
    print(error)
```
